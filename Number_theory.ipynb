## 01


```python

def extended_gcd(a, b):
    assert a >= b and b >= 0 and a + b > 0

    d, x, y = 0, 0, 0

    # YOUR CODE HERE
    if b == 0:
        d = a
        x = 1
        y = 0
    else:
        d, x1, y1 = extended_gcd(b, a % b)
        x = y1
        y = x1 - (a // b) * y1

    assert a % d == 0 and b % d == 0
    assert d == a * x + b * y

    return (d, x, y)


print(extended_gcd(8, 4))
print(extended_gcd(27, 15))
```

## 02

```python

def lcm(a, b):
    # Find GCD using Euclidean Algorithm
    x, y = a, b

    while y != 0:
        x, y = y, x % y

    gcd = x

    # Calculate LCM
    return (a * b) // gcd
```

```python
assert lcm(9, 3) == 9
assert lcm(17, 37) == 629

print(lcm(9, 3))
print(lcm(17, 37))
```

## 03

```python

def diophantine(a, b, c):
    # Calculate GCD and coefficients
    if a >= b:
        gcd, x1, y1 = extended_gcd(a, b)
    else:
        gcd, y1, x1 = extended_gcd(b, a)

    # Calculate x and y
    x, y = 0, 0

    # YOUR CODE HERE
    factor = c // gcd

    x = x1 * factor
    y = y1 * factor

    assert c % gcd == 0
    assert a * x + b * y == c

    return (x, y)


print(diophantine(34, 51, 170))
print(diophantine(91, 117, 65))
```

## 04

```python

def modInverse(a, n):
    """Calculate the multiplicative inverse of a (mod n)"""

    if a >= n:
        gcd, x, y = extended_gcd(a, n)
    else:
        gcd, y, x = extended_gcd(n, a)

    return x % n


def divide(b, a, n):
    assert n > 1 and a > 0

    aInv = modInverse(a, n)

    assert (a * aInv) % n == 1

    # return the number x s.t. a*x = b (mod n)
    return (aInv * b) % n


print(divide(9, 5, 3))
print(divide(223, 67, 54))
```

```python

def FastModularExponentiation(b, e, m):
    result = 1
    b = b % m

    while e > 0:
        # If exponent is odd
        if e % 2 == 1:
            result = (result * b) % m

        # Square the base
        b = (b * b) % m

        # Divide exponent by 2
        e = e // 2

    return result


print(FastModularExponentiation(4, 5, 3))
```

```python

def ChineseRemainderTheorem(n1, r1, n2, r2):

    # Calculate Extended GCD
    (d, x, y) = extended_gcd(n1, n2)

    assert d == 1

    # YOUR CODE HERE
    k = ((r2 - r1) * x) % n2

    result = (r1 + n1 * k) % (n1 * n2)

    return result


print(ChineseRemainderTheorem(13, 7, 6, 5))  # 59
print(ChineseRemainderTheorem(28, 20, 19, 3))  # 440
```
